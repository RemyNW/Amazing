```python
import os
import pandas as pd
import gzip
from pathlib import Path

# Définir les chemins
gz_dir = Path('../data/gz')
parquet_dir = Path('../data/duck_db')
parquet_dir.mkdir(parents=True, exist_ok=True)

# Lister tous les fichiers .csv.gz dans le dossier gz
gz_files = list(gz_dir.glob('*.csv.gz'))

for gz_file in gz_files:
    # Définir le nom du fichier de sortie Parquet
    parquet_file = parquet_dir / gz_file.with_suffix('').with_suffix('.parquet').name
    if parquet_file.exists():
        print(f"[SKIP] {parquet_file.name} existe déjà.")
        continue
    print(f'Conversion de {gz_file.name}...')
    # Lire le fichier CSV.gz en DataFrame
    with gzip.open(gz_file, 'rt') as f:
        df = pd.read_csv(f)
    # Sauvegarder en Parquet
    df.to_parquet(parquet_file, index=False)
    print(f'Fichier converti : {parquet_file}')

print('Conversion terminée.')
```
